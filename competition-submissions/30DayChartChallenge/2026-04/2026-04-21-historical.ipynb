```python
"""
#30DayChartChallenge — Day 21: Historical
"The Population Explosion: 500 Years of Human Growth" — long-term historical timeseries
in the style of FlowingData: minimal chrome, direct labels, clean typography.

Data: Global population from 1500 to 2023
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------------------------------
# 1. SAMPLE DATA — World population over 500 years (Billions)
# ---------------------------------------------------------------------------
data = {
    "Year": [1500, 1600, 1700, 1800, 1850, 1900, 1920, 1940, 1950, 1960, 1970, 1980, 1990, 2000, 2010, 2023],
    "Population": [0.46, 0.58, 0.68, 0.98, 1.26, 1.65, 1.86, 2.30, 2.52, 3.04, 3.70, 4.44, 5.30, 6.13, 6.96, 8.08]
}

df = pd.DataFrame(data)

# ---------------------------------------------------------------------------
# 2. PLOT — Historical population curve with annotations
# ---------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(14, 8), facecolor="#FFFFFF")
ax.set_facecolor("#FFFFFF")

# Main line
ax.plot(df["Year"], df["Population"], color="#d62728", linewidth=4, marker="o", 
        markersize=8, zorder=3, markerfacecolor="white", markeredgewidth=2.5)

# Fill area under curve
ax.fill_between(df["Year"], 0, df["Population"], color="#d62728", alpha=0.15, zorder=1)

# Highlight key historical eras with background shading
ax.axvspan(1500, 1800, alpha=0.08, color="#2ca02c", zorder=0)  # Pre-industrial
ax.axvspan(1800, 1900, alpha=0.08, color="#ff7f0e", zorder=0)  # Industrial Revolution
ax.axvspan(1900, 1950, alpha=0.08, color="#1f77b4", zorder=0)  # 20th century rise
ax.axvspan(1950, 2023, alpha=0.08, color="#9467bd", zorder=0)  # Modern explosion

# Era labels
ax.text(1650, 7.5, "Pre-Industrial", fontsize=9, color="#2ca02c", fontweight="600", ha="center", alpha=0.7)
ax.text(1850, 7.5, "Industrial\nRevolution", fontsize=9, color="#ff7f0e", fontweight="600", ha="center", alpha=0.7)
ax.text(1925, 7.5, "20th Century", fontsize=9, color="#1f77b4", fontweight="600", ha="center", alpha=0.7)
ax.text(1987, 7.5, "Modern Era", fontsize=9, color="#9467bd", fontweight="600", ha="center", alpha=0.7)

# Key milestone annotations
milestones = [
    (1800, 0.98, "1 Billion"),
    (1927, 2.0, "2 Billion"),
    (1960, 3.04, "3 Billion"),
    (1974, 4.0, "4 Billion"),
    (1987, 5.0, "5 Billion"),
    (1999, 6.0, "6 Billion"),
    (2011, 7.0, "7 Billion"),
    (2023, 8.08, "8 Billion")
]

for year, pop, label in milestones:
    ax.plot(year, pop, marker="o", markersize=10, color="#d62728", zorder=5, 
            markerfacecolor="#fff", markeredgewidth=2.5)
    ax.text(year, pop + 0.35, label, fontsize=9, fontweight="bold", color="#d62728", ha="center",
            bbox=dict(boxstyle="round,pad=0.3", facecolor="#fff", edgecolor="#d62728", linewidth=1, alpha=0.9))

# Key historical events
events = [
    (1665, 0.5, "Great Plague\nof London"),
    (1918, 1.8, "Spanish\nFlu"),
    (1945, 2.35, "WWII\nEnds"),
    (1969, 3.6, "Moon\nLanding")
]

for year, pop, event in events:
    ax.annotate("", xy=(year, pop - 0.15), xytext=(year, pop - 0.35),
                arrowprops=dict(arrowstyle="-|>", color="#888888", lw=1, alpha=0.4))
    ax.text(year, pop - 0.5, event, fontsize=7, color="#888888", ha="center", style="italic", alpha=0.6)

# Minimal styling
ax.set_xlim(1480, 2030)
ax.set_ylim(0, 8.5)
ax.spines[["top", "right"]].set_visible(False)
ax.spines[["left", "bottom"]].set_color("#cccccc")
ax.grid(True, color="#eeeeee", linewidth=0.8, alpha=0.6, axis="y", zorder=0)
ax.set_axisbelow(True)

ax.set_xlabel("Year", fontsize=11, color="#666666", fontweight="500")
ax.set_ylabel("Population (Billions)", fontsize=11, color="#666666", fontweight="500")
ax.set_xticks([1500, 1600, 1700, 1800, 1900, 2000, 2023])
ax.set_yticks([0, 2, 4, 6, 8])
ax.tick_params(colors="#888888", labelsize=10)

# Title & subtitle
ax.text(0.02, 0.97, "The Population Explosion: 500 Years of Human Growth",
        fontsize=22, fontweight="bold", color="#1a1a1a", transform=ax.transAxes, va="top")

ax.text(0.02, 0.90,
        "Global population grew slowly for millennia, then exploded after industrialization.\n"
        "From 0.46B in 1500 to 8.08B in 2023—most growth occurred in just the last 70 years.\n"
        "The curve is flattening as birth rates decline in developed nations.",
        fontsize=10.5, color="#666666", transform=ax.transAxes, va="top", linespacing=1.5)

# Growth rate context
ax.text(0.98, 0.25, "Doubling Times\n1500→1800: 300 yrs\n1800→1900: 100 yrs\n1950→2023: 73 yrs",
        fontsize=8.5, color="#333333", transform=ax.transAxes, ha="right", va="center",
        bbox=dict(boxstyle="round,pad=0.5", facecolor="#f9f9f9", edgecolor="#d62728", linewidth=1.5, alpha=0.9))

# Source
ax.text(0.02, 0.01,
        "Data: Historical world population estimates (UN, World Bank, Worldometer 1500-2023) · #30DayChartChallenge Day 21: Historical",
        fontsize=9, color="#aaaaaa", style="italic", transform=ax.transAxes, va="bottom")

plt.tight_layout()
plt.savefig("day21_historical.png", dpi=200, facecolor="#FFFFFF", bbox_inches="tight")
plt.show()
```
