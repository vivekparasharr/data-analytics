```python
"""
#30DayChartChallenge — Day 12: Theme Day (Reporters Without Borders)
"Press Freedom Under Pressure" — global press freedom index
in the style of FlowingData: minimal chrome, direct labels, clean typography.

Data: Reporters Without Borders Press Freedom Index (2023)
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------------------------------
# 1. SAMPLE DATA — Press Freedom Index by country (0-100, higher is better)
# ---------------------------------------------------------------------------
data = {
    "Country": [
        "Norway",
        "Ireland",
        "Denmark",
        "Sweden",
        "Costa Rica",
        "Netherlands",
        "Jamaica",
        "Lithuania",
        "New Zealand",
        "Switzerland",
        "United States",
        "Germany",
        "Canada",
        "France",
        "Japan",
        "Mexico",
        "India",
        "Brazil",
        "Russia",
        "China"
    ],
    "Freedom_Index": [
        94.0, 93.5, 93.0, 92.5, 91.0,
        90.5, 90.0, 88.5, 88.0, 87.5,
        83.0, 82.5, 80.0, 78.5, 76.0,
        55.0, 48.0, 45.5, 25.0, 22.0
    ],
    "Status": [
        "Good", "Good", "Good", "Good", "Good",
        "Good", "Good", "Good", "Good", "Good",
        "Satisfactory", "Satisfactory", "Satisfactory", "Satisfactory", "Satisfactory",
        "Problematic", "Problematic", "Problematic", "Very Poor", "Very Poor"
    ]
}

df = pd.DataFrame(data)
df = df.sort_values("Freedom_Index", ascending=True)

# ---------------------------------------------------------------------------
# 2. PLOT
# ---------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(12, 9), facecolor="#FFFFFF")
ax.set_facecolor("#FFFFFF")

# Color by status
color_map = {
    "Good": "#2ca02c",
    "Satisfactory": "#ff7f0e",
    "Problematic": "#d62728",
    "Very Poor": "#8b0000"
}

colors = [color_map[status] for status in df["Status"]]

# Horizontal bar chart
bars = ax.barh(range(len(df)), df["Freedom_Index"], height=0.65, 
               color=colors, edgecolor="white", linewidth=1, zorder=3)

# Add value labels
for i, (idx, row) in enumerate(df.iterrows()):
    ax.text(row["Freedom_Index"] + 1.5, i, f"{row['Freedom_Index']:.1f}",
            fontsize=9, fontweight="600", color="#333333", va="center")

# Y-axis labels
ax.set_yticks(range(len(df)))
ax.set_yticklabels(df["Country"], fontsize=10, fontweight="500", color="#333333")

# Minimal styling
ax.set_xlim(0, 105)
ax.spines[["top", "right", "left"]].set_visible(False)
ax.spines["bottom"].set_color("#cccccc")
ax.xaxis.grid(True, color="#eeeeee", linewidth=0.8, alpha=0.6, zorder=0)
ax.set_axisbelow(True)
ax.set_xticks([0, 25, 50, 75, 100])
ax.set_xticklabels(["0", "25", "50", "75", "100"], fontsize=9, color="#888888")

# Title & subtitle
ax.text(0.02, 0.97, "Press Freedom Under Pressure",
        fontsize=22, fontweight="bold", color="#1a1a1a", transform=ax.transAxes, va="top")

ax.text(0.02, 0.91,
        "Reporters Without Borders Press Freedom Index (2023). Higher scores indicate greater press freedom.\n"
        "Democratic nations dominate the top, while authoritarian regimes rank lowest.",
        fontsize=10.5, color="#666666", transform=ax.transAxes, va="top", linespacing=1.5)

# Legend
legend_y = 0.85
ax.text(0.02, legend_y, "■ Good", fontsize=9, color="#2ca02c", transform=ax.transAxes, fontweight="600")
ax.text(0.15, legend_y, "■ Satisfactory", fontsize=9, color="#ff7f0e", transform=ax.transAxes, fontweight="600")
ax.text(0.35, legend_y, "■ Problematic", fontsize=9, color="#d62728", transform=ax.transAxes, fontweight="600")
ax.text(0.58, legend_y, "■ Very Poor", fontsize=9, color="#8b0000", transform=ax.transAxes, fontweight="600")

# Annotation
ax.text(0.98, 0.15, "Free Press\nMatters",
        fontsize=11, fontweight="bold", color="#2ca02c", transform=ax.transAxes, 
        ha="right", va="center", bbox=dict(boxstyle="round,pad=0.5",
        facecolor="#fff", edgecolor="#2ca02c", linewidth=2, alpha=0.95))

# Source
ax.text(0.02, 0.01,
        "Source: Reporters Without Borders Press Freedom Index (2023) · #30DayChartChallenge Day 12: Theme Day",
        fontsize=9, color="#aaaaaa", style="italic", transform=ax.transAxes, va="bottom")

plt.tight_layout()
plt.savefig("day12_press_freedom.png", dpi=200, facecolor="#FFFFFF", bbox_inches="tight")
plt.show()
```
