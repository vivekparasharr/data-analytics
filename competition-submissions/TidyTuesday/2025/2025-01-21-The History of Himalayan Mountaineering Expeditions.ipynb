```python
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-01-21/peaks_tidy.csv"
peaks = pd.read_csv(url)

status_counts = (
    peaks.groupby(["HIMAL_FACTOR", "PSTATUS_FACTOR"])
    .size()
    .reset_index(name="count")
)

plt.figure(figsize=(14, 8))
sns.barplot(
    data=status_counts,
    y="HIMAL_FACTOR",
    x="count",
    hue="PSTATUS_FACTOR",
    palette="Set2"
)

plt.title("Climbing Status by Himalayan Mountain Range")
plt.xlabel("Number of Peaks")
plt.ylabel("Mountain Range")
plt.legend(title="Climbing Status")
plt.tight_layout()
plt.show()
```
