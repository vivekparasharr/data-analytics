```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import numpy as np

# Load the datasets
water_quality = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-05-20/water_quality.csv')
weather = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-05-20/weather.csv')

# Convert date columns to datetime
water_quality['date'] = pd.to_datetime(water_quality['date'])
weather['date'] = pd.to_datetime(weather['date'])

# Set seaborn style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = [14, 8]

# Chart 1: Enterococci levels over time with rainfall overlay
fig, ax1 = plt.subplots(figsize=(14, 7))

# Plot enterococci levels
ax1.scatter(water_quality['date'], water_quality['enterococci_cfu_100ml'], 
           alpha=0.3, color='blue', label='Enterococci (CFU/100ml)', s=20)
ax1.set_xlabel('Date', fontsize=12)
ax1.set_ylabel('Enterococci Levels (CFU/100ml)', color='blue', fontsize=12)
ax1.tick_params(axis='y', labelcolor='blue')

# Create secondary axis for rainfall
ax2 = ax1.twinx()
rainfall_data = weather.groupby('date')['precipitation_mm'].sum().reset_index()
ax2.bar(rainfall_data['date'], rainfall_data['precipitation_mm'], 
       alpha=0.3, color='green', label='Rainfall (mm)', width=2)
ax2.set_ylabel('Rainfall (mm)', color='green', fontsize=12)
ax2.tick_params(axis='y', labelcolor='green')

plt.title('Sydney Beach Water Quality and Rainfall Over Time\n(1991-2025)', fontsize=14, fontweight='bold')
fig.tight_layout()
plt.legend(loc='upper left')
plt.xticks(rotation=45)
plt.savefig('water_quality_vs_rainfall.png', dpi=300, bbox_inches='tight')
plt.show()

# Chart 2: Boxplot of enterococci by rainfall categories
weather_merged = water_quality.merge(weather[['date', 'precipitation_mm']], on='date', how='left')
weather_merged['precipitation_mm'] = weather_merged['precipitation_mm'].fillna(0)

# Create rainfall categories
weather_merged['rainfall_category'] = pd.cut(weather_merged['precipitation_mm'], 
                                            bins=[-0.1, 0, 5, 10, 20, 50, 100, float('inf')],
                                            labels=['No Rain', '<5mm', '5-10mm', '10-20mm', '20-50mm', '50-100mm', '>100mm'])

plt.figure(figsize=(12, 6))
sns.boxplot(data=weather_merged, x='rainfall_category', y='enterococci_cfu_100ml', 
           order=['No Rain', '<5mm', '5-10mm', '10-20mm', '20-50mm', '50-100mm', '>100mm'])
plt.xlabel('Rainfall Category', fontsize=12)
plt.ylabel('Enterococci Levels (CFU/100ml)', fontsize=12)
plt.title('Enterococci Levels by Rainfall Amount\n(Sydney Beaches)', fontsize=14, fontweight='bold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('enterococci_by_rainfall.png', dpi=300, bbox_inches='tight')
plt.show()

# Chart 3: Heatmap of average enterococci by beach and rainfall
beach_rainfall_pivot = weather_merged.groupby(['swim_site', 'rainfall_category'])['enterococci_cfu_100ml'].mean().unstack()

plt.figure(figsize=(14, 8))
sns.heatmap(beach_rainfall_pivot, annot=True, fmt='.1f', cmap='YlOrRd', 
           linewidths=0.5, cbar_kws={'label': 'Avg Enterococci (CFU/100ml)'})
plt.title('Average Enterococci Levels by Beach and Rainfall\n(Correlation Analysis)', fontsize=14, fontweight='bold')
plt.xlabel('Rainfall Category', fontsize=12)
plt.ylabel('Beach/Swimming Site', fontsize=12)
plt.tight_layout()
plt.savefig('beach_rainfall_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()

# Chart 4: Scatter plot with trend line
plt.figure(figsize=(12, 6))
sns.scatterplot(data=weather_merged, x='precipitation_mm', y='enterococci_cfu_100ml', 
               alpha=0.4, hue='region', palette='Set2', s=50)
sns.regplot(data=weather_merged, x='precipitation_mm', y='enterococci_cfu_100ml', 
           scatter=False, color='red', line_kws={'linewidth': 2})
plt.xlabel('Precipitation (mm)', fontsize=12)
plt.ylabel('Enterococci Levels (CFU/100ml)', fontsize=12)
plt.title('Relationship Between Rainfall and Bacteria Levels\n(with Trend Line)', fontsize=14, fontweight='bold')
plt.legend(title='Region', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('rainfall_enterococci_scatter.png', dpi=300, bbox_inches='tight')
plt.show()

print("Charts generated successfully!")
print(f"Dataset loaded: {len(water_quality)} water quality records, {len(weather)} weather records")
print(f"Date range: {water_quality['date'].min()} to {water_quality['date'].max()}")
```
