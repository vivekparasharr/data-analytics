```python
import pandas as pd
import matplotlib.pyplot as plt

base = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-12-23/"

languages = pd.read_csv(base + "languages.csv")
statuses = pd.read_csv(base + "endangered_status.csv")

# Join each language to its recorded status
data = languages.merge(statuses, on="id", how="inner")

# Count languages with a recorded status in each macroarea
counts = (
    data.dropna(subset=["macroarea"])
        .groupby("macroarea")
        .size()
        .sort_values()
)

# Plot
ax = counts.plot(kind="barh", figsize=(9, 5), color="steelblue")
ax.set_title("Languages with Recorded Endangerment Status by Macroarea")
ax.set_xlabel("Number of languages")
ax.set_ylabel("Macroarea")

plt.tight_layout()
plt.show()
```
