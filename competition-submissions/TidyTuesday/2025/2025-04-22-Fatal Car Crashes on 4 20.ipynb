```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Set style for cleaner charts
sns.set_theme(style="whitegrid")

# 1. Load the dataset directly from the TidyTuesday GitHub repository
url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-04-22/daily_accidents_420.csv"
df_420 = pd.read_csv(url)

# Convert the 'date' column to datetime objects
df_420["date"] = pd.to_datetime(df_420["date"])

# 2. Filter for instances where e420 is True (accidents occurring during the 4/20 holiday window)
df_holiday = df_420[df_420["e420"] == True].copy()
df_holiday["year"] = df_holiday["date"].dt.year

# 3. Create the Visualization
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=df_holiday,
    x="year",
    y="fatalities_count",
    marker="o",
    color="#2ca02c",  # Forest green accent
    linewidth=2.5,
    markersize=6,
)

plt.title(
    "Fatal Car Crashes During the 4/20 Window (4:20 PM – 11:59 PM on April"
    " 20th)",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Fatalities Count", fontsize=12)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.show()
```
