```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load the dataset directly from the TidyTuesday GitHub repository
url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-05-27/monsters.csv"
monsters = pd.read_csv(url)

# Quick check of the data structure
print(f"Loaded {len(monsters)} monsters successfully!")
print(monsters[["name", "type", "cr", "hp_number"]].head())

# 2. Prepare the data for plotting
# Filter for the top 8 most common monster types to keep the chart clean and readable
top_types = monsters["type"].value_counts().head(8).index
filtered_monsters = monsters[monsters["type"].isin(top_types)]

# 3. Create the visualization
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Generate a box plot of Challenge Ratings by Monster Type
sns.boxplot(
    data=filtered_monsters,
    x="type",
    y="cr",
    hue="type",
    palette="viridis",
    legend=False,
    linewidth=1.5,
)

# Customize chart details
plt.title(
    "Challenge Rating (CR) Distribution Across Top D&D Monster Types (2024 SRD)",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Monster Type", fontsize=12, fontweight="bold")
plt.ylabel("Challenge Rating (CR)", fontsize=12, fontweight="bold")
plt.xticks(rotation=30, ha="right", fontsize=10)
plt.yticks(fontsize=10)

# Adjust layout and display
plt.tight_layout()
plt.show()
```
